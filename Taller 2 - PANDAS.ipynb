```python
import pandas as pd
df_mundial = pd.read_csv(r"/content/WorldCupMatches-selected-columns.csv")
df_mundial.head()
```

```python
df_mundial.info()
```

Eliminación de duplicados

```python
df_mundial.duplicated().sum()
```

```python
df_mundial[df_mundial.duplicated(keep=False)]
```

```python
df_mundial.drop_duplicates(inplace=True, keep="first")
df_mundial.duplicated().sum()
```

```python
df_mundial = df_mundial.reset_index(drop=True)
df_mundial.head()
```

Limpieza datos numericos

```python
df_mundial["Datetime"] = pd.to_datetime(df_mundial["Datetime"], errors="coerce")
df_mundial.info()
df_mundial.head()
```

```python
df_mundial["Year"] = df_mundial["Year"].astype(int)
df_mundial["Home Team Goals"] = df_mundial["Home Team Goals"].astype(int)
df_mundial["Away Team Goals"] = df_mundial["Away Team Goals"].astype(int)
df_mundial.info()
df_mundial.head()
```

1.¿Cuales son las 6 ciudades que mas partidos han visto disputarse y cuantos partidos en cada una?

```python
sorted(df_mundial["City"].unique())
```

```python
df_mundial["City"] = df_mundial["City"].str.strip()
Reemplazos_ciudades = {"Berlin West":"Berlin",
              "New York/New Jersey":"New Jersey"}
df_mundial["City"] = df_mundial["City"].replace(Reemplazos_ciudades)
sorted(df_mundial["City"].unique())
```

```python
ciudades = pd.DataFrame(df_mundial["City"].value_counts())
ciudades.head(6)
```

2.¿Cuales son los 3 paises que mas partidos a jugado en los mundiales?

```python
sorted(df_mundial["Home Team Name"].unique())
```

```python
Reemplazos_paises = {"Czech Republic":"Czechia",
              "Côte d'Ivoire":"C\u00f4te d'Ivoire",
              "Ivory Coast":"C\u00f4te d'Ivoire",
              "German DR":"Germany",
              "Germany FR":"Germany",
              "IR Iran":"Iran",
              "Korea DPR":"Korea Republic",
              "Türkiye":"Turkey",
              "USA":"United States",
              "rn\">Bosnia and Herzegovina":"Bosnia and Herzegovina",
              "rn\">Republic of Ireland":"Republic of Ireland",
              "rn\">Serbia and Montenegro":"Serbia and Montenegro",
              "rn\">Trinidad and Tobago":"Trinidad and Tobago",
              "rn\">United Arab Emirates":"United Arab Emirates",
              "Soviet Union":"Russia"}
```

```python
df_mundial["Home Team Name"] = df_mundial["Home Team Name"].str.strip()
df_mundial["Home Team Name"] = df_mundial["Home Team Name"].replace(Reemplazos_paises)

sorted(df_mundial["Home Team Name"].unique())
```

```python
sorted(df_mundial["Away Team Name"].unique())
```

```python
df_mundial["Away Team Name"] = df_mundial["Away Team Name"].str.strip()
df_mundial["Away Team Name"] = df_mundial["Away Team Name"].replace(Reemplazos_paises)

sorted(df_mundial["Away Team Name"].unique())
```

```python
df_paises = pd.concat([df_mundial["Home Team Name"], df_mundial["Away Team Name"]])
df_paises.value_counts().head(3)
```

# los paises que màs partidos han jugado en los mundiales son alemania, brasil y argentina

3.¿Que pais ha jugado mas finales sin ganarlas?

```python
sorted(df_mundial["Stage"].unique())
```

```python
df_mundial["Stage"] = df_mundial["Stage"].str.strip()
Reemplazos_finales = {"Finals":"Final", "final":"Final"}
df_mundial["Stage"] = df_mundial["Stage"].replace(Reemplazos_finales)
sorted(df_mundial["Stage"].unique())
```

```python
df_finalistas = df_mundial[df_mundial['Stage'] == "Final"]
df_finalistas
```

```python
df_finalistas_locales_perdedores = df_finalistas["Home Team Goals"] < df_finalistas["Away Team Goals"]
df_finalistas_locales_perdedores
```

```python
df_finalistas_visitantes_perdedores = df_finalistas["Away Team Goals"] < df_finalistas["Home Team Goals"]
df_finalistas_visitantes_perdedores
```

```python
df_equipos_locales_perdedores = df_finalistas[df_finalistas_locales_perdedores]['Home Team Name']
df_equipos_locales_perdedores
```

```python
df_equipos_visitantes_perdedores = df_finalistas[df_finalistas_visitantes_perdedores]['Away Team Name']
df_equipos_visitantes_perdedores
```

```python
df_finalistas_perdedores = pd.concat([df_equipos_locales_perdedores, df_equipos_visitantes_perdedores])
df_finalistas_perdedores.value_counts()
```

# Los paises que mas finales han perdido son Alemania y argentina.
